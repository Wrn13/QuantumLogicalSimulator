```python
import qutip as qt

from quantum_logical.gates import sqrtISWAP_operator, iSWAP_operator

ge_gate = iSWAP_operator(3,1)
ef_gate = iSWAP_operator(3,2)

combined_gate = ef_gate*ge_gate

for i in range(3):
    for j in range(3):
        state = qt.tensor(qt.basis(3,i), qt.basis(3,j))
        print(f"For state {i}{j}, gate result: {combined_gate * state}") 
```
