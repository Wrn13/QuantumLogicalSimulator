```python
import sys
import numpy as np
np.set_printoptions(threshold=sys.maxsize)
```

```python
from quantum_logical.channel import AmplitudeDamping
import qutip as qt

gammas=[.25, .5, .75, .9]
trotter_dt = 1
for gamma in gammas:

    T1 = -1 * trotter_dt / np.log(1-gamma)
    amp_damp = AmplitudeDamping(T1, 2, 2)
    amp_damp.set_trotter_dt(1)
    plus = qt.snot() * qt.basis(2,0)
    rho = qt.ket2dm(qt.tensor(plus, plus))

    print(amp_damp(rho))

    print("E_00 = ", np.kron(np.array([[1, 0],[0, np.sqrt(1-gamma)]], np.eye(4)))
    print("E_1 = ", np.array([[0, np.sqrt(gamma)],[0, 0]]))


```
